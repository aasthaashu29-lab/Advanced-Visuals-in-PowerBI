# **Advanced Visuals in PowerBI**



# 1. Difference Between Basic Filter, Advanced Filter, and Slicer in Power BI

| Feature       | Basic Filter                         | Advanced Filter                                     | Slicer                                     |
| ------------- | ------------------------------------ | --------------------------------------------------- | ------------------------------------------ |
| Purpose       | Filters data using simple conditions | Filters data using complex conditions               | Interactive visual filter for users        |
| Usage         | Select values from a list            | Use conditions like contains, greater than, between | Users click options directly on the report |
| Example       | Show sales for 2023 only             | Show sales greater than ₹50,000                     | Filter report by Region                    |
| Interactivity | Static                               | Static                                              | Dynamic and interactive                    |
| Location      | Filter Pane                          | Filter Pane                                         | Report Canvas                              |

### Summary

* **Basic Filter** is used for selecting simple values.
* **Advanced Filter** is used when logical conditions are needed.
* **Slicer** improves user interaction by allowing live filtering directly on the dashboard.

# 2. Apply a Basic Filter to Display Sales Data for the Year 2023

## Steps

1. Open Power BI Desktop.
2. Import the `Sales_dataset`.
3. Add any visual (Table/Chart).
4. Go to the **Filters Pane**.
5. Drag the **Year** field into the filter section.
6. Select **Basic Filtering**.
7. Check only **2023**.

### Result

All visuals now display only sales data for the year 2023.

# 3. Create a Pie Chart of Sales by Product and Add a Region Slicer

## Steps

### Create Pie Chart

   * **Product** → Legend
   * **Sales** → Values

### Add Slicer

1. Select the **Slicer** visual.
2. Drag **Region** into the slicer field.
3. Choose dropdown or list format.

### Result

Users can select a region and the pie chart updates automatically.

# 4. Difference Between Visual-Level, Page-Level, and Report-Level Filters

| Filter Type         | Applies To              | Example                             |
| ------------------- | ----------------------- | ----------------------------------- |
| Visual-Level Filter | One visual only         | Filter one chart to show 2023 sales |
| Page-Level Filter   | All visuals on one page | Show only India data on Page 1      |
| Report-Level Filter | Entire report           | All pages show India sales only     |

### Summary

* Visual-level filters affect only one chart.
* Page-level filters affect all visuals on a single page.
* Report-level filters affect the entire Power BI report.

---

# 5. Apply a Report-Level Filter for India Sales Only

## Steps

1. Open the Filters Pane.
2. Locate **Report-Level Filters**.
3. Drag the **Country** field into it.
4. Select only **India**.

### Result

All report pages and visuals now display sales data only for India.

# 6. Role of Drill-Down in Power BI

## Explanation

Drill-Down allows users to move from summarized data to detailed data within a hierarchy.

### Benefits

* Improves data exploration
* Helps identify trends and patterns
* Makes reports interactive
* Enables multi-level analysis

### Example

A user can analyze:

* Country sales
* Then Region sales
* Then Product Category sales

without creating separate visuals.

---

# 7. Create a Hierarchy Drill-Down: Country → Region → Product Category

## Steps

1. Create a chart (Bar/Column Chart).
2. Add hierarchy fields in this order:

   * Country
   * Region
   * Product Category
3. Enable **Drill Mode** from the visual header.
4. Click on chart elements to drill deeper.

### Result

Users can navigate:
Country → Region → Product Category

and analyze sales at multiple levels.

# 8. Implement Drill-Through for Product-Wise Region Sales

## Steps

### Create Drill-Through Page

1. Add a new report page.
2. Name it: `Product Details`.
3. Drag **Product** into the **Drill-through Filters** area.

### Add Visuals

1. Add a table or chart.
2. Include:

   * Region
   * Sales

### Use Drill-Through

1. Right-click a product from another page.
2. Select **Drill-through → Product Details**.

### Result

Users navigate to a detailed page showing region-wise sales for the selected product.

# 10. Difference Between Drill-Down and Drill-Through

| Feature    | Drill-Down               | Drill-Through                     |
| ---------- | ------------------------ | --------------------------------- |
| Navigation | Within same visual       | Opens another report page         |
| Purpose    | Explore hierarchy levels | View detailed analysis            |
| Example    | Year → Quarter → Month   | Product → Detailed Product Report |

## Real-World Example

### Drill-Down

A sales manager views:

* Country sales
* Then drills into Region sales
* Then drills into Product Category sales

### Drill-Through

The manager right-clicks on a product and opens a separate page showing:

* Monthly sales
* Regional performance
* Customer details

# 13. What is “Keep All Filters” in Drill-Through?

## Explanation

The **Keep All Filters** option preserves all currently applied filters when moving to a drill-through page.

### Example

If users apply:

* Region = North
* Year = 2023

and then drill-through into Product Details, those filters remain active.

### Benefits

* Maintains report context
* Improves user experience
* Provides consistent analysis
